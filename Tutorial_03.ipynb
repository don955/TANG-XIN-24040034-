```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# ==========================================
# 1. Load dataset & Visualize
# ==========================================
df = sns.load_dataset('titanic')

print("--- Question 1: Dataset Overview ---")
display(df.head())

plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='class', hue='survived')
plt.title('Survival Count by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.show()

# ==========================================
# 2, 3, 4. Samples & Features Evaluation
# ==========================================
num_samples = df.shape[0]
num_features = df.shape[1]

print("\n--- Question 2 & 4: Samples ---")
print(f"Total Samples (Rows): {num_samples}")
print("Evaluation: Yes, 891 samples are sufficient for exploratory analysis and baseline machine learning models.")

print("\n--- Question 3: Features ---")
print(f"Total Features (Columns): {num_features}")
print("Evaluation: Yes, 15 features provide enough information including demographics, ticket info, and family counts.")

# ==========================================
# 5. Datatypes
# ==========================================
print("\n--- Question 5: Datatypes ---")
print(df.dtypes)

# ==========================================
# 6 & 8. Missing / Null Values Analysis
# ==========================================
print("\n--- Question 6 & 8: Null / Missing Values ---")
null_summary = df.isnull().sum()
print("Columns with null values and counts:")
print(null_summary[null_summary > 0])

missing_rows = df[df.isnull().any(axis=1)].index.tolist()
print(f"\nTotal rows with missing values: {len(missing_rows)}")
print(f"Row indices with missing values (first 10 rows): {missing_rows[:10]}")

# ==========================================
# 7. Target vs Features
# ==========================================
print("\n--- Question 7: Target and Features ---")
print("Target Output Column: 'survived' (0 = Did not survive, 1 = Survived)")
features_list = [col for col in df.columns if col != 'survived']
print(f"Feature Columns: {features_list}")

# ==========================================
# 9. Research Question & Suitability
# ==========================================
print("\n--- Question 9: Question & Suitability ---")
print("Question: What socio-economic factors (e.g., passenger class, age, gender) were most critical in predicting survival?")
print("Suitability: Yes, the dataset is fully suitable because it contains both feature variables and ground-truth survival labels.")
```
